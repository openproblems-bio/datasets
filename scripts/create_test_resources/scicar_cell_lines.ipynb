# Test resources: sci-CAR cell lines

sci-CAR jointly measures the transcriptome and chromatin accessibility of the same cells using combinatorial indexing ([Cao et al., 2018](https://doi.org/10.1126/science.aau0730)). This dataset contains 4,825 cells from three cell lines (human HEK293T and A549 cells, and mouse NIH/3T3 cells), collected 0, 1 and 3 hours after dexamethasone treatment.

```python
import anndata as ad
import lamindb as ln

import helpers

helpers.check_instance()
ln.track("Aq0TwdUdIrHt", project=helpers.PROJECT)
helpers.link_to_github("scripts/create_test_resources/scicar_cell_lines.ipynb")
```

## Select cells and features

We subsample the dataset to 600 cells and 1500 features per modality.

## Process

Each modality is normalised in the way that suits it: gene expression to log counts per 10,000 (log CP10k), protein abundance (ADT) with a centred log-ratio transformation (CLR) and chromatin accessibility (ATAC) with TF-IDF. For each modality we then compute a truncated SVD embedding and select highly variable features.

```python
output_dir = helpers.run_workflow("multimodal/process_openproblems_v1_multimodal", {
    "id": "scicar_cell_lines",
    "input_id": "scicar_cell_lines",
    "obs_tissue": "source",
    "layer_counts": "counts",
    "obs_cell_type": "cell_name",
    "var_feature_id": "index",
    "var_feature_name": "gene_short_name",
    "dataset_name": "sci-CAR cell lines",
    "dataset_url": "https://www.ncbi.nlm.nih.gov/geo/query/acc.cgi?acc=GSE117089",
    "dataset_reference": "cao2018joint",
    "dataset_summary": "sciCAR is a combinatorial indexing-based assay that jointly measures cellular transcriptomes and the accessibility of cellular chromatin in the same cells",
    "dataset_description": "sciCAR is a combinatorial indexing-based assay that jointly measures cellular transcriptomes and the accessibility of cellular chromatin in the same cells. Here, we use two sciCAR datasets that were obtained from the same study. The first dataset contains 4,825 cells from three cell lines (HEK293T cells, NIH/3T3 cells, and A549 cells) at multiple timepoints (0, 1 hour, 3 hours) after dexamethasone treatment. The second dataset contains 11,233 cells from wild-type adult mouse kidney.",
    "dataset_organism": "[homo_sapiens, mus_musculus]",
    "mod1": "GEX",
    "mod2": "ATAC",
    "do_subsample": True,
    "n_obs": 600,
    "n_vars": 1500,
    "seed": 123,
    "normalization_methods": ["log_cp10k"],
    "output_mod1": "$id/dataset_mod1.anndata.zarr",
    "output_mod2": "$id/dataset_mod2.anndata.zarr",
    "output_state": "$id/state.yaml",
})
```
